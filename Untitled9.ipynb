```python
def factorial(n):
    if n == 0:
        return 1
    else:
        res = 1
        for i in range(1, n + 1):
            res *= i
        return res

num = int(input("Enter a non-negative number to find its factorial: "))

if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    print(f"The factorial of {num} is {factorial(num)}")
```

### Alphabet Triangle
This code will print an alphabet triangle pattern.

```python
n = 5 # Number of rows for the triangle

for i in range(n):
    for j in range(i + 1):
        print(chr(65 + j), end="") # 65 is the ASCII value for 'A'
    print()
```

### Alphabet Triangle (Alternate Method without ASCII)
This method uses string slicing to print the alphabet triangle.

```python
alphabets = "ABCDEF"
n = 5 # Number of rows for the triangle

for i in range(n):
    print(alphabets[:i+1])
```
